```python

PROJECT_DIR = Path.cwd().parent

RAW_BOOKING_PATH = (
    PROJECT_DIR
    / "data"
    / "raw"
    / "booking_data.csv"
)

PROCESSED_DATA_DIR = (
    PROJECT_DIR
    / "data"
    / "processed"
)

booking_df = pd.read_csv(RAW_BOOKING_PATH)

booking_df.head()
```

```python
# Remove adress

hotels_cleaned_df = booking_df.drop(
    columns=["address"]
)
```

```python
# Check duplicates
hotels_cleaned_df.duplicated(
    subset=["name", "city"]
).sum()

```

```python
# Save
hotels_cleaned_df.to_csv(
    PROCESSED_DATA_DIR / "hotels_cleaned.csv",
    index=False
)
```
